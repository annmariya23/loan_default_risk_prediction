```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("LoanDefault") \
    .master("local[*]") \
    .getOrCreate()

df = spark.read.parquet(data_path)

print("Total rows:", df.count())
df.printSchema()
```

```python
from pyspark.sql.functions import col

# Keep only useful ML columns
feature_cols = [
    "loan_amnt",
    "funded_amnt",
    "funded_amnt_inv",
    "int_rate",
    "installment",
    "annual_inc",
    "dti",
    "delinq_2yrs",
    "inq_last_6mths",
    "open_acc",
    "pub_rec",
    "revol_bal",
    "revol_util",
    "total_acc",
    "mort_acc",
    "pub_rec_bankruptcies"
]

ml_df = df.select(feature_cols + ["default_flag"])

# Remove rows with missing target
ml_df = ml_df.filter(col("default_flag").isNotNull())

print("Rows available:", ml_df.count())
```

```python
ml_df.groupBy("default_flag").count().show()
```

```python
from pyspark.sql.functions import col

# Remove rows where the target or features are missing
ml_df = ml_df.dropna()

print("Rows after removing missing values:", ml_df.count())

# Use 20% of the cleaned data for model training/testing
# 1.35 million rows -> roughly 270k rows
sample_df = ml_df.sample(withReplacement=False, fraction=0.20, seed=42)

print("Sample rows:", sample_df.count())

# Split into training and testing
train_data, test_data = sample_df.randomSplit([0.8, 0.2], seed=42)

print("Training rows:", train_data.count())
print("Testing rows:", test_data.count())
```

```python
from pyspark.ml.feature import VectorAssembler

feature_cols = [
    "loan_amnt",
    "funded_amnt",
    "funded_amnt_inv",
    "int_rate",
    "installment",
    "annual_inc",
    "dti",
    "delinq_2yrs",
    "inq_last_6mths",
    "open_acc",
    "pub_rec",
    "revol_bal",
    "revol_util",
    "total_acc",
    "mort_acc",
    "pub_rec_bankruptcies"
]

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="features",
    handleInvalid="skip"
)

train_vec = assembler.transform(train_data).select("features", "default_flag")
test_vec = assembler.transform(test_data).select("features", "default_flag")

print("Training data ready:", train_vec.count())
print("Testing data ready:", test_vec.count())
```

```python
from pyspark.ml.classification import LogisticRegression

lr = LogisticRegression(
    featuresCol="features",
    labelCol="default_flag",
    maxIter=20
)

lr_model = lr.fit(train_vec)

lr_predictions = lr_model.transform(test_vec)

print("Logistic Regression completed!")
```

```python
from pyspark.ml.classification import RandomForestClassifier

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="default_flag",
    numTrees=50,
    maxDepth=8,
    seed=42
)

rf_model = rf.fit(train_vec)

rf_predictions = rf_model.transform(test_vec)

print("Random Forest completed!")
```

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# -----------------------------
# Logistic Regression metrics
# -----------------------------

lr_evaluator = BinaryClassificationEvaluator(
    labelCol="default_flag",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

lr_auc = lr_evaluator.evaluate(lr_predictions)

# F1
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

f1_evaluator = MulticlassClassificationEvaluator(
    labelCol="default_flag",
    predictionCol="prediction",
    metricName="f1"
)

accuracy_evaluator = MulticlassClassificationEvaluator(
    labelCol="default_flag",
    predictionCol="prediction",
    metricName="accuracy"
)

lr_f1 = f1_evaluator.evaluate(lr_predictions)
lr_accuracy = accuracy_evaluator.evaluate(lr_predictions)


# -----------------------------
# Random Forest metrics
# -----------------------------

rf_auc = lr_evaluator.evaluate(rf_predictions)

rf_f1 = f1_evaluator.evaluate(rf_predictions)
rf_accuracy = accuracy_evaluator.evaluate(rf_predictions)


# -----------------------------
# Display comparison
# -----------------------------

print("=" * 55)
print("MODEL COMPARISON")
print("=" * 55)

print("\nLogistic Regression")
print("Accuracy :", round(lr_accuracy, 4))
print("F1 Score :", round(lr_f1, 4))
print("AUC      :", round(lr_auc, 4))

print("\nRandom Forest")
print("Accuracy :", round(rf_accuracy, 4))
print("F1 Score :", round(rf_f1, 4))
print("AUC      :", round(rf_auc, 4))


# -----------------------------
# Select better model
# -----------------------------

if rf_f1 > lr_f1:
    best_model = rf_model
    best_name = "Random Forest"
    best_f1 = rf_f1
else:
    best_model = lr_model
    best_name = "Logistic Regression"
    best_f1 = lr_f1

print("\n" + "=" * 55)
print("BEST MODEL")
print("=" * 55)
print("Selected:", best_name)
print("F1 Score:", round(best_f1, 4))
```

FINAL MODEL SELECTION

Logistic Regression was selected as the final model.

Although Random Forest achieved a slightly higher accuracy,
Logistic Regression achieved a higher F1 Score and AUC.
The accuracy difference between the two models was very small.

Therefore, Logistic Regression was selected as the final
model for loan default prediction.

```python
from pyspark.sql import Row
from pyspark.sql.functions import when, col

# ============================================================
# 3 SAMPLE LOAN APPLICANTS
# ============================================================

applicants = [
    # Applicant 1 - Lower risk profile
    Row(
        applicant="Applicant 1",
        loan_amnt=10000.0,
        funded_amnt=10000.0,
        funded_amnt_inv=10000.0,
        int_rate=8.0,
        installment=300.0,
        annual_inc=75000.0,
        dti=12.0,
        delinq_2yrs=0.0,
        inq_last_6mths=0.0,
        open_acc=12.0,
        pub_rec=0.0,
        revol_bal=3000.0,
        revol_util=20.0,
        total_acc=25.0,
        mort_acc=2.0,
        pub_rec_bankruptcies=0.0
    ),

    # Applicant 2 - Medium risk profile
    Row(
        applicant="Applicant 2",
        loan_amnt=20000.0,
        funded_amnt=20000.0,
        funded_amnt_inv=20000.0,
        int_rate=13.5,
        installment=500.0,
        annual_inc=50000.0,
        dti=22.0,
        delinq_2yrs=1.0,
        inq_last_6mths=2.0,
        open_acc=8.0,
        pub_rec=0.0,
        revol_bal=9000.0,
        revol_util=55.0,
        total_acc=15.0,
        mort_acc=1.0,
        pub_rec_bankruptcies=0.0
    ),

    # Applicant 3 - Higher risk profile
    Row(
        applicant="Applicant 3",
        loan_amnt=30000.0,
        funded_amnt=30000.0,
        funded_amnt_inv=30000.0,
        int_rate=20.0,
        installment=800.0,
        annual_inc=35000.0,
        dti=35.0,
        delinq_2yrs=3.0,
        inq_last_6mths=5.0,
        open_acc=4.0,
        pub_rec=1.0,
        revol_bal=18000.0,
        revol_util=85.0,
        total_acc=8.0,
        mort_acc=0.0,
        pub_rec_bankruptcies=1.0
    )
]

# Create Spark DataFrame
new_loans = spark.createDataFrame(applicants)

# Keep only the model features for prediction
new_loans_features = new_loans.select(
    "applicant",
    *feature_cols
)

# Convert features into vector
new_loans_vec = assembler.transform(new_loans_features)

# Make prediction using FINAL LOGISTIC REGRESSION MODEL
predictions = lr_model.transform(new_loans_vec)

# Add readable risk result
final_prediction = predictions.withColumn(
    "Risk",
    when(col("prediction") == 1.0, "HIGH RISK")
    .otherwise("LOW RISK")
)

# Show final results
final_prediction.select(
    "applicant",
    "prediction",
    "probability",
    "Risk"
).show(truncate=False)
```
